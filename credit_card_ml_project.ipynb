# Model Building

## Logistic Regression

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, recall_score, precision_score, f1_score, precision_recall_curve, classification_report
```

```python
lr = LogisticRegression(max_iter = 600)
lr.fit(X_train, y_train)
```

```python
y_pred_lr = lr.predict(X_test)
```

```python
print(f"\n========== Logistic Regression ==========")
print(f"\nAccuracy :  {accuracy_score(y_test, y_pred_lr)}")
print(f"\n classification_report :\n  {classification_report(y_test, y_pred_lr)}")
```

```python
from sklearn.metrics import confusion_matrix
confusion_matrix(y_test, y_pred_lr)
```

```python
from sklearn.metrics import roc_curve
```

```python
y_score = lr.predict_proba(X_test)[:, 1]         # probability of having 1 (fraud)
y_score
```

```python
fpr, tpr, thresholds = roc_curve(y_test, y_score)
```

```python
# roc_curve
import plotly.express as px
fig = px.line(x= fpr, y= tpr, title= 'roc-curve', labels = {'x' : 'FPR', 'y' : 'TPR'}, width= 700, height= 500)
fig.add_shape(type= 'line', line = {'dash': 'dash', 'color': 'gray'}, x0= 0, y0= 0, x1= 1, y1= 1)
fig.show()
```

```python
#optimum threshold value
best_threshold1 = thresholds[np.argmax(tpr-fpr)]
print(best_threshold1)
y_pred_lr_optimum = (y_score >= best_threshold1).astype(int)
```

```python
print(f"\n========== Logistic Regression ==========\n")
print(f"\nAccuracy :  {accuracy_score(y_test, y_pred_lr_optimum)}\n")

print(f"Optimal Threshold (F1): {best_threshold1:.3f}\n")

# 5. Print updated metrics
print(classification_report(y_test, y_pred_lr_optimum))
```

```python
from sklearn.metrics import precision_recall_curve
precision, recall, thresholds = precision_recall_curve(y_test, y_score)
f1_scores = 2*(precision * recall) / (precision+  recall + 1e-8)

best_f1_score = np.argmax(f1_scores)
best_threshold = thresholds[best_f1_score]

y_pred_lr_optimum2 = (y_score >= best_threshold).astype(int)

print(f"\n========== Logistic Regression ==========\n")
print(f"\nAccuracy :  {accuracy_score(y_test, y_pred_lr_optimum2)}\n")
print(f"Optimal Threshold : {best_threshold:.3f}\n")

print(classification_report(y_test, y_pred_lr_optimum2))
```

```python
lr2 = LogisticRegression(class_weight = 'balanced', max_iter = 1000)
lr2.fit(X_train, y_train)
```

```python
y_score2 = lr2.predict_proba(X_test)[:, 1]         # probability of having 1 (fraud)
```

```python
precision, recall, thresholds = precision_recall_curve(y_test, y_score2)
f1_scores = 2*(precision * recall) / (precision+  recall + 1e-8)
best_f1_score = np.argmax(f1_scores)
best_threshold2 = thresholds[best_f1_score]

y_pred_lr_optimum2 = (y_score2 >= best_threshold2).astype(int)

print(f"\n========== Logistic Regression ==========\n")
print(f"\nAccuracy :  {accuracy_score(y_test, y_pred_lr_optimum2)}\n")
print(f"Optimal Threshold2 : {best_threshold2:.3f}\n")

print(classification_report(y_test, y_pred_lr_optimum2))
```

## Best model among Logistic Regression [maintain imbalanced data]

```python
from sklearn.calibration import CalibratedClassifierCV

base_lr = LogisticRegression(class_weight='balanced', max_iter=1400)

calibrated_lr = CalibratedClassifierCV(estimator=base_lr, method='sigmoid', cv=5)
calibrated_lr.fit(X_train, y_train)

y_probs_calibrated = calibrated_lr.predict_proba(X_test)[:, 1]

precision, recall, thresholds = precision_recall_curve(y_test, y_probs_calibrated)
f1_scores = 2 * (precision * recall) / (precision + recall + 1e-8)

best_threshold = thresholds[np.argmax(f1_scores)]
print(f"\n========== Logistic Regression ==========\n")

print(f"\nNew Calibrated Threshold: {best_threshold:}\n")

y_pred = (y_probs_calibrated >= best_threshold).astype(int)
print(f"classification_report : \n {classification_report(y_test, y_pred)}")
```

## Random Forest Classifier

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
y_pred_rf = rf.predict(X_test)
```

### Threshold Tuning [this is the best model among all Random Forest Classifier]

# FINAL MODEL

```python
rf = RandomForestClassifier(class_weight = 'balanced_subsample',
                            n_estimators=100,
                            max_depth=10,
                            max_samples=0.7,
                            oob_score= True,
                            n_jobs=-1,
                            random_state=42)
rf.fit(X_train, y_train)
```

```python
y_pred_proba = rf.predict_proba(X_test)[:, 1]
```

```python
precision, recall, thresholds = precision_recall_curve(y_test, y_pred_proba)
f1_scores = 2*(precision * recall)/(precision + recall + 1e-8)

best_f1_score = np.argmax(f1_scores)

best_threshold = thresholds[best_f1_score]

y_pred_optimum = (y_pred_proba >= best_threshold).astype(int)


print(f"\n========== Random Forest Classifier ==========\n")
print(f"\nAccuracy :  {accuracy_score(y_test, y_pred_optimum)}\n")
print(f"Optimal Threshold : {best_threshold:.4f}\n")

print(classification_report(y_test, y_pred_optimum))
```

```python
from sklearn.model_selection import cross_val_score, StratifiedKFold
```

```python
print(cross_val_score(rf, X_train, y_train, cv= 5, scoring= 'accuracy').mean())
print(cross_val_score(rf, X_train, y_train, cv= 5, scoring= 'f1').mean())
```
